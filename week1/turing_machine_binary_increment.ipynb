# Turing Machine: Binary Incrementer

This notebook implements a simple **Turing Machine** that increments a binary number by 1.

## Problem
Given a binary string (e.g. `1011`), the Turing Machine produces its incremented value (`1100`).

We assume:
- The tape initially contains only `0` and `1`
- The head starts on the **rightmost bit**
- The machine halts when the increment is complete


## Turing Machine Definition

States:
- `q0`: Add 1 (handle carry)
- `qhalt`: Halt state

Alphabet:
- `0`, `1`, `_` (blank)

Transition logic:
- If reading `1`, write `0` and move left (carry continues)
- If reading `0`, write `1` and halt (carry resolved)
- If reading `_`, write `1` and halt (overflow case)


```
class TuringMachine:
    def __init__(self, tape, head_position, transitions, start_state, halt_states):
        self.tape = list(tape)
        self.head = head_position
        self.state = start_state
        self.transitions = transitions
        self.halt_states = halt_states

    def step(self):
        symbol = self.tape[self.head] if self.head < len(self.tape) else '_'
        if self.state in self.halt_states:
            return False

        key = (self.state, symbol)
        if key not in self.transitions:
            raise RuntimeError(f"No transition defined for {key}")

        new_state, write_symbol, move = self.transitions[key]

        # Write
        if self.head < len(self.tape):
            self.tape[self.head] = write_symbol
        else:
            self.tape.append(write_symbol)

        # Move
        if move == 'L':
            self.head -= 1
            if self.head < 0:
                self.tape.insert(0, '_')
                self.head = 0
        elif move == 'R':
            self.head += 1

        self.state = new_state
        return True

    def run(self, max_steps=100):
        steps = 0
        while steps < max_steps and self.step():
            steps += 1
        return ''.join(self.tape).strip('_')
```

## Transition Function

```
transitions = {
    ('q0', '1'): ('q0', '0', 'L'),   # carry continues
    ('q0', '0'): ('qhalt', '1', 'N'), # carry resolved
    ('q0', '_'): ('qhalt', '1', 'N'), # overflow (e.g. 111 + 1)
}
```

## Example: Increment a Binary Number

```
# Input binary number
binary_input = "1011"  # 11 in decimal

# Initialize the Turing Machine
tm = TuringMachine(
    tape=binary_input,
    head_position=len(binary_input) - 1,
    transitions=transitions,
    start_state='q0',
    halt_states={'qhalt'}
)

# Run the machine
result = tm.run()

print(f"Input:  {binary_input}")
print(f"Output: {result}")
```

## Result

The Turing Machine correctly increments the binary number.

`1011 → 1100`

This demonstrates how a simple Turing Machine can perform arithmetic using only symbolic manipulation.
